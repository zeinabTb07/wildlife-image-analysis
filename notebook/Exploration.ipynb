# فاز دوم: استخراج ویژگی و تحلیل تصاویر

در فاز اول، داده‌های مشاهدات جانوری ایران از iNaturalist جمع‌آوری و پاک‌سازی شدند. تصاویر نیز دانلود و از نظر کیفیت، محتوای بصری و موارد تکراری بررسی شدند. این بررسی‌ها مشکلاتی مانند عدم توازن گونه‌ها، تفاوت کیفیت و محتوای تصاویر، وجود عکس‌های نامرتبط و نبود برچسب مشخص برای وضعیت ظاهری حیوانات (مانند آسیب‌دیدگی یا مرگ) را آشکار کرد.

در این فاز، با استفاده از مدل‌های بینایی ماشین و استخراج بردارهای ویژگی (Embedding)، ساختار محتوایی تصاویر بررسی می‌شود. همچنین روش‌های کاهش ابعاد و خوشه‌بندی برای شناسایی الگوهای تصویری به کار گرفته می‌شوند و از مدل‌های هوش مصنوعی آفلاین برای تحلیل و بررسی محتوای تصاویر و نمونه‌های مبهم استفاده می‌شود.
